## OpenCV Uses BGR Color Format

OpenCV stores images and video frames in BGR (Blue, Green, Red) channel order instead of RGB.

This design choice comes from OpenCV’s origins in C/C++, where BGR was commonly used in low-level image processing libraries.

Each frame read using OpenCV is a NumPy array with shape (height, width, 3).

The three channels represent Blue, Green, and Red intensity values in that order.

Many deep learning models and visualization libraries expect RGB format, so direct usage of BGR images may cause incorrect color interpretation.

To ensure correct visualization or model input, OpenCV images are usually converted from BGR to RGB using cv2.cvtColor().

## Matplotlib Uses RGB Format

Matplotlib uses the RGB color format for displaying images, where pixel values are arranged in the order of Red, Green, and Blue channels. This format is widely accepted in visualization libraries and deep learning frameworks. When an image is displayed using plt.imshow(), Matplotlib assumes the input image follows RGB channel ordering. However, since OpenCV reads images in BGR format, displaying them directly in Matplotlib results in incorrect color representation. Therefore, it is necessary to convert images from BGR to RGB before visualization to ensure accurate color rendering.

```python
import cv2
fig, axs = plt.subplots(5, 5, figsize=(30, 20))
axs = axs.flatten()

cap = cv2.VideoCapture("026c7465-309f6d33.mp4")
n_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

img_idx = 0
for frame in range(n_frames):
    ret, img = cap.read()
    if ret == False:
        break
    if frame % 100 == 0:                                          #frame is divisible by 100 - Note the modulo operator
        axs[img_idx].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        axs[img_idx].set_title(f'Frame: {frame}')
        axs[img_idx].axis('off')
        img_idx += 1

plt.tight_layout()                                                #automatically adjusts the spacing between the plots
plt.show()
cap.release()
```

## Video Quality Dimensions – Resolution Levels


###  1. **480p (SD – Standard Definition)**

* Resolution: **854 × 480 pixels**
* Low image clarity and detail
* Requires less storage and bandwidth
* Suitable for basic monitoring where fine details are not required
* Limited accuracy for object detection in surveillance

###  2. **720p (HD – High Definition)**

* Resolution: **1280 × 720 pixels**
* Moderate image quality with clearer object shapes
* Balanced choice between quality and storage
* Commonly used in entry-level surveillance cameras
* Suitable for basic vehicle and person detection



###  3. **1080p (Full HD)**

* Resolution: **1920 × 1080 pixels**
* High clarity and sharp details
* Enables accurate object detection and recognition
* Widely used in modern CCTV and traffic surveillance
* Requires higher storage and processing power



###  4. **1440p (2K / Quad HD)**

* Resolution: **2560 × 1440 pixels**
* Very high image detail and sharpness
* Suitable for large-area surveillance and zoom-based analysis
* Improves detection accuracy for distant objects
* Requires high bandwidth, storage, and GPU resources




## What are Annotations?

Annotations are labels or markings added to images or videos to describe the content.

They provide ground truth information used for training and evaluating machine learning models.

In computer vision, annotations identify what object is present, where it is located, and what action is happening.

Common types of annotations include class labels, bounding boxes, segmentation masks, and keypoints.

In video data, annotations may also include temporal information, such as start and end frames of an activity.

Annotations are essential for supervised learning, especially in object detection, tracking, and activity recognition tasks.

## How Annotations Are Used in Surveillance

Annotations are used to label objects such as vehicles, people, or suspicious items in surveillance videos.

They provide ground truth data for training object detection and tracking models.

Bounding box annotations help systems locate and track moving objects across frames.

Temporal annotations define when an event starts and ends, such as intrusion or loitering.

Annotated data improves the accuracy of behavior and activity recognition systems.

Surveillance systems use annotations to generate alerts when predefined rules are violated.

Annotations support video analytics, such as counting vehicles or monitoring restricted areas.

## How Annotations Improve Surveillance Systems

Annotations provide accurate ground truth data, which helps surveillance models learn correct object and event patterns.

They improve the accuracy of object detection, such as identifying cars, people, or intrusions.

Annotated videos enable reliable object tracking across multiple frames and cameras.

Temporal annotations help systems understand event duration, such as loitering or unauthorized entry.

High-quality annotations reduce false alarms by teaching the system what is normal and abnormal behavior.

Annotations support activity recognition, allowing systems to detect suspicious or illegal actions.

They help in building automated alert systems for real-time surveillance monitoring.
